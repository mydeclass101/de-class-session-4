```python
import sys

print(sys.executable)          # ต้องมีคำว่า .venv อยู่ใน path
print(sys.version)
assert sys.version_info >= (3, 11), "ต้องใช้ Python 3.11 ขึ้นไป"
if ".venv" not in sys.executable:
    print("⚠️ kernel ไม่ใช่ .venv ของ repo นี้ ถ้า cell ถัดไป error ให้เปลี่ยน kernel (มุมขวาบน)")
print("✅ Python พร้อม")
```

```python
import boto3
import duckdb
import pandas as pd
import psycopg
import requests
import sqlalchemy
from dotenv import load_dotenv

print("pandas", pd.__version__, "| duckdb", duckdb.__version__, "| boto3", boto3.__version__)
print("✅ library ครบ")
```

```python
import os

found_env = load_dotenv()
print("เจอไฟล์ .env:", found_env)
assert found_env, "ยังไม่มีไฟล์ .env (คัดลอกจาก .env.example)"
print("✅ .env พร้อม")
```

```python
api = os.getenv("SHOP_API_URL", "http://127.0.0.1:8000")
print(requests.get(f"{api}/health", timeout=5).json())
print("✅ API ร้านค้าพร้อม")
```

```python
db = {
    "host": os.getenv("POSTGRES_HOST", "127.0.0.1"),
    "port": os.getenv("POSTGRES_PORT", "5432"),
    "dbname": os.getenv("POSTGRES_DB", "ecommerce"),
    "user": os.getenv("POSTGRES_USER", "student"),
    "password": os.getenv("POSTGRES_PASSWORD", "student123"),
}
with psycopg.connect(**db) as conn:
    print("orders:", conn.execute("SELECT count(*) FROM orders").fetchone()[0])
print("✅ PostgreSQL ร้านค้าพร้อม")
```
