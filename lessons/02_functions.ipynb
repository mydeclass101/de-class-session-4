# บทที่ 2 · Function

function = **ชุดคำสั่งที่ตั้งชื่อไว้ เรียกใช้ซ้ำได้** เหมือน "สูตรอาหาร": เขียนครั้งเดียว ทำกี่ครั้งก็ได้

บทนี้: **def · return · parameter · ค่า default · keyword argument · type hint · ส่ง function เป็นค่า · decorator**

## 2.1 สร้างและเรียกใช้ function

````python
def make_greeting(name):
    """สร้างคำทักทาย"""                      # docstring: อธิบายว่า function นี้ทำอะไร
    return f"สวัสดีครับคุณ{name}"


message = make_greeting("มานี")              # เรียกใช้: ชื่อ + วงเล็บ + ค่าที่ส่งเข้าไป
print(message)
print(make_greeting("สมชาย"))
````

- `def ชื่อ(parameter):` แล้วย่อหน้าเนื้อใน
- `return` = **ส่งผลลัพธ์กลับ** ให้คนที่เรียก (และจบ function ทันที)

⚠️ `print` ≠ `return` : `print` แค่แสดงบนจอ ส่วน `return` ส่งค่ากลับไปให้ใช้ต่อได้

````python
def show_price(price):
    print(f"ราคา {price} บาท")         # แค่แสดง ไม่ได้ส่งอะไรกลับ


result = show_price(55)
print("ได้ค่ากลับมา:", result)         # None: function ไม่มี return
````

## 2.2 parameter หลายตัว, ค่า default, keyword argument

````python
def calculate_bill(price, qty, discount=0):
    """ราคารวม หลังหักส่วนลด (discount เป็นทศนิยม เช่น 0.1 = 10%)"""
    return price * qty * (1 - discount)


print(calculate_bill(55, 2))                        # ไม่ส่ง discount -> ใช้ค่า default 0
print(calculate_bill(55, 2, 0.1))                   # ส่งตามลำดับ
print(calculate_bill(qty=2, price=55, discount=0.1))   # ส่งแบบระบุชื่อ (keyword) สลับลำดับได้ อ่านง่ายกว่า
````

💡 เวลาเรียก function ที่มี parameter เยอะ ๆ นิยมส่งแบบ **ชื่อ=ค่า** (keyword argument) อ่านแล้วรู้เลยว่าแต่ละค่าคืออะไร

```python
book_table(restaurant="ส้มตำป้าแดง", people=4, time="19:00", need_baby_chair=False)
```

## 2.3 Type hint: บอกว่าควรส่งอะไรเข้า และได้อะไรกลับ

`name: str` = ควรเป็นข้อความ · `-> dict` = function นี้คืนค่าเป็น dict
Python **ไม่ได้บังคับ** (ส่งผิดก็ไม่ error ทันที) แต่ช่วยให้คนอ่านเข้าใจ และ VS Code ช่วยเตือนได้

````python
def make_receipt(menu: str, qty: int, price: float) -> dict:
    return {"menu": menu, "qty": qty, "total": qty * price}


receipt = make_receipt("latte", 2, 55)
print(receipt, receipt["total"])
````

💡 function ที่คืน **dict** มีประโยชน์มาก: ส่งผลลัพธ์หลายอย่างกลับมาพร้อมกัน และมีชื่อกำกับทุกค่า

## 2.4 รับค่ากี่ตัวก็ได้: `*args` และ `**kwargs`

- `*args` : รวมค่าที่ส่งมาเกินเป็น **tuple** (คล้าย list)
- `**kwargs` : รวม `ชื่อ=ค่า` ที่ส่งมาเกินเป็น **dict**

อ่านออกก็พอ เห็นบ่อยในโค้ดของคนอื่น

````python
def total_price(*prices):
    print("ได้รับ:", prices)
    return sum(prices)


print(total_price(45, 12, 38))
print(total_price(100))
````

````python
def order_drink(menu, **options):
    print("เมนู:", menu)
    print("ตัวเลือกเพิ่ม:", options)            # options เป็น dict
    print("หวาน:", options.get("sweet", "ปกติ"))


order_drink("ชาเย็น", sweet="น้อย", ice="มาก", size="L")
````

````python
# ทิศกลับกัน: ** หน้า dict = "แตก dict ออกเป็น ชื่อ=ค่า" ตอนเรียก function
my_usual = {"price": 55, "qty": 2, "discount": 0.1}
print(calculate_bill(**my_usual))            # เท่ากับ calculate_bill(price=55, qty=2, discount=0.1)
````

## 2.5 function ก็เป็น "ค่า" ส่งต่อให้ function อื่นได้

ชื่อ function **ไม่มีวงเล็บ** = ตัว function (สูตรอาหาร)
ชื่อ function **มีวงเล็บ** = เรียกใช้แล้ว ได้ผลลัพธ์ (อาหารที่ทำเสร็จ)

````python
def boil_egg():
    return "ไข่ต้ม"


print(boil_egg)        # <function boil_egg ...>  ตัว function
print(boil_egg())      # 'ไข่ต้ม'                  ผลจากการเรียก
````

````python
def cook_for_family(recipe, people):
    """รับ "สูตร" เข้ามา แล้วทำตามสูตรให้ครบทุกคน"""
    return [recipe() for _ in range(people)]


print(cook_for_family(boil_egg, 3))      # ส่งตัว function เข้าไป (ไม่มีวงเล็บ)
````

⚠️ ผิดบ่อย: `cook_for_family(boil_egg(), 3)` (มีวงเล็บ) = ต้มไข่ **ทันที** แล้วส่ง "ไข่" เข้าไปแทน "สูตร" → error

## 2.6 Decorator: `@ชื่อ` ที่อยู่เหนือ function

decorator คือ **function ที่รับ function เข้าไป แล้วคืน function ที่ "ทำได้มากขึ้น" ออกมา**
โดยไม่ต้องแก้เนื้อในของ function เดิม เหมือน **ห่อของขวัญ**: ของข้างในเหมือนเดิม แต่มีกระดาษห่อเพิ่ม

```python
@wrap_gift              มีความหมายเดียวกับ      def make_cake(): ...
def make_cake(): ...                              make_cake = wrap_gift(make_cake)
```

**ไม่ต้องฝึกเขียน decorator เอง** แค่ดูตัวอย่างข้างล่างให้เข้าใจว่า `@` ทำอะไร

````python
import time


def timer(func):
    """decorator: จับเวลาว่า function ทำงานนานเท่าไหร่"""
    def wrapper(*args, **kwargs):
        print(f"เริ่ม {func.__name__}")
        start = time.time()
        result = func(*args, **kwargs)                 # เรียก function ตัวจริง
        print(f"เสร็จ {func.__name__} ใช้เวลา {time.time() - start:.2f} วินาที")
        return result
    return wrapper


@timer
def make_coffee(menu):
    time.sleep(0.5)                                    # สมมติว่าชงครึ่งวินาที
    return f"{menu} พร้อมเสิร์ฟ"


print(make_coffee("latte"))          # function เดิม แต่มีการจับเวลาเพิ่มให้อัตโนมัติ
````

### decorator ที่รับค่าได้: `@try_again(times=3)`

ขั้นแรก `try_again(times=3)` สร้าง decorator ออกมาก่อน แล้วค่อยเอาไปห่อ function

````python
def try_again(times=3):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, times + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as problem:
                    print(f"ครั้งที่ {attempt} ไม่สำเร็จ: {problem}")
            raise RuntimeError(f"{func.__name__} ไม่สำเร็จทั้ง {times} ครั้ง")
        return wrapper
    return decorator


call_count = {"n": 0}


@try_again(times=3)
def call_friend():
    call_count["n"] += 1
    if call_count["n"] < 3:
        raise ConnectionError("สายไม่ว่าง")
    return "เพื่อนรับสายแล้ว"


print(call_friend())
````

💡 สรุป decorator ในประโยคเดียว: **เขียน function ธรรมดา แล้วแปะ `@...` เพื่อเพิ่มความสามารถ** เช่น จับเวลา ลองใหม่ บันทึก log
library หลายตัวให้ decorator สำเร็จรูปมาใช้ เราแค่แปะไว้บน function ของเรา

## 2.7 lambda: function สั้น ๆ ไม่มีชื่อ (อ่านออกก็พอ)

````python
double = lambda x: x * 2                 # เท่ากับ def double(x): return x * 2
print(double(21))

menus = [{"menu": "latte", "price": 55}, {"menu": "mocha", "price": 65}, {"menu": "espresso", "price": 45}]
print(sorted(menus, key=lambda m: m["price"]))      # เรียงตามราคา
````

## ✅ สรุปบทที่ 2

| เรื่อง | ตัวอย่าง |
|---|---|
| สร้าง / เรียก | `def make_greeting(name): return ...` → `make_greeting("มานี")` |
| default / keyword | `def bill(price, qty, discount=0)` → `bill(price=55, qty=2)` |
| type hint | `def make_receipt(menu: str, qty: int) -> dict:` |
| `*args` / `**kwargs` | รับค่ากี่ตัวก็ได้ / รับ `ชื่อ=ค่า` กี่ตัวก็ได้ (เป็น dict) |
| function เป็นค่า | `cook_for_family(boil_egg, 3)` ไม่มีวงเล็บ |
| decorator | `@timer`, `@try_again(times=3)` เพิ่มความสามารถโดยไม่แก้ function เดิม |
