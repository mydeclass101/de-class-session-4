## 1.2 ตัวแปร (variable) และชนิดข้อมูล

ตัวแปร = **ชื่อ** ที่ใช้เก็บ **ค่า** · สร้างด้วย `=` · ตั้งชื่อเป็นภาษาอังกฤษตัวเล็ก คั่นคำด้วย `_`

```python
coffee_name = "latte"          # str   ข้อความ (อยู่ในเครื่องหมาย " " หรือ ' ')
coffee_price = 55              # int   จำนวนเต็ม
sugar_level = 0.5              # float ทศนิยม
is_iced = True                 # bool  จริง/เท็จ (True / False ตัวแรกพิมพ์ใหญ่)
note = None                    # None  "ยังไม่มีค่า"

print(coffee_name, coffee_price, sugar_level, is_iced, note)
print(type(coffee_name), type(coffee_price), type(sugar_level), type(is_iced), type(note))
```

```python
# คำนวณได้ตามปกติ
cups = 3
total = coffee_price * cups
print("จ่ายทั้งหมด", total, "บาท")
print(100 / 3, 100 // 3, 100 % 3, 2 ** 10)      # หาร, หารปัดเศษทิ้ง, เศษ, ยกกำลัง
```

⚠️ `"55"` (ข้อความ) ไม่ใช่ `55` (ตัวเลข) ข้อมูลที่อ่านจากไฟล์หรือจากเว็บมักมาเป็น **ข้อความ** ต้องแปลงก่อนคำนวณ

```python
price_from_file = "55"
print(price_from_file + "5")           # ต่อข้อความ -> '555'
print(int(price_from_file) + 5)        # แปลงเป็นตัวเลขก่อน -> 60
print(float("12.50"), str(99) + " บาท")
```

## 1.3 ข้อความ (string) และ f-string

**f-string** = ข้อความที่มี `f` นำหน้า ใส่ตัวแปรใน `{ }` ได้เลย ใช้บ่อยที่สุดในการสร้างข้อความ

```python
name = "สมชาย"
age = 25
print(f"{name} อายุ {age} ปี")
print(f"กาแฟ {cups} แก้ว ราคา {total:,} บาท")        # :, = ใส่ comma คั่นหลักพัน
print(f"ส่วนลด {0.15:.0%}")                            # :.0% = แสดงเป็นเปอร์เซ็นต์
```

```python
# สิ่งที่ทำกับข้อความได้บ่อย ๆ (เรียกว่า method: ชื่อ.คำสั่ง())
menu_item = "  Iced Latte  "
print(menu_item.strip())                    # ตัดช่องว่างหัวท้าย
print(menu_item.strip().lower())            # ตัวพิมพ์เล็ก
print(menu_item.strip().upper())            # ตัวพิมพ์ใหญ่
print("photo_2026.jpg".endswith(".jpg"))    # ลงท้ายด้วย...ไหม
print("25/09/2026".split("/"))              # แยกเป็นหลายชิ้น
print("-".join(["2026", "09", "25"]))       # รวมหลายชิ้นเป็นข้อความเดียว
print("ชาเย็น".replace("เย็น", "ร้อน"))
```

```python
# ข้อความหลายบรรทัด ใช้ """ ... """
receipt = f"""
ร้านกาแฟหน้าออฟฟิศ
{coffee_name} x {cups}
รวม {total} บาท
"""
print(receipt)
```

## 1.4 List: รายการของหลายค่า (มีลำดับ)

```python
shopping_list = ["ไข่", "นม", "ขนมปัง"]

print(shopping_list[0])          # ตัวแรก (นับจาก 0)
print(shopping_list[-1])         # ตัวสุดท้าย
print(len(shopping_list))        # มีกี่ชิ้น
print("นม" in shopping_list)     # มีนมในรายการไหม

shopping_list.append("กาแฟ")     # เพิ่มท้ายรายการ
print(shopping_list)
print(shopping_list[1:3])        # ชิ้นที่ 1 ถึง 2
```

## 1.5 Dict: เก็บค่าเป็นคู่ "ชื่อ → ค่า"

💡 dict เป็นโครงสร้างที่ใช้บ่อยที่สุดในงานข้อมูล: 1 dict = ข้อมูล 1 รายการ (1 แถว)

```python
friend = {
    "name": "มานี",
    "age": 24,
    "city": "ขอนแก่น",
}

print(friend["name"])              # อ่านค่าด้วยชื่อ (key)
friend["age"] = 25                 # แก้ค่า
friend["phone"] = "081-234-5678"   # เพิ่ม key ใหม่
print(friend)
```

```python
# อ่าน key ที่อาจไม่มี: ใช้ .get(key, ค่าสำรอง) จะไม่ error
print(friend.get("email"))                   # None
print(friend.get("email", "ไม่มีอีเมล"))
print("city" in friend)
```

```python
friend["email"]      # ❌ KeyError: ไม่มี key นี้  (อ่านบรรทัดสุดท้ายของ error ก่อนเสมอ)
```

```python
# วนดูทุกคู่
for key, value in friend.items():
    print(key, "=", value)
```

## 1.6 dict + list ซ้อนกัน

ข้อมูลจริงมักซ้อนกันแบบนี้ เช่น "ออเดอร์ 1 ใบ มีสินค้าหลายชิ้น"
(ข้อมูลจากเว็บ/แอปเกือบทุกที่หน้าตาแบบนี้ เรียกว่า **JSON**)

```python
order = {
    "order_id": "A001",
    "customer": "มานี",
    "items": [
        {"menu": "latte", "qty": 2, "price": 55},
        {"menu": "croissant", "qty": 1, "price": 65},
    ],
}

print(order["customer"])
print(order["items"][0])                # สินค้าชิ้นแรก (เป็น dict)
print(order["items"][0]["menu"])        # ชื่อสินค้าชิ้นแรก
print(len(order["items"]), "รายการ")
```

## 1.7 if / elif / else

⚠️ Python ใช้ **การย่อหน้า 4 ช่อง** บอกว่าบรรทัดไหนอยู่ในเงื่อนไข

```python
money = 120

if money >= 200:
    print("ซื้อกาแฟ + เค้กได้")
elif money >= 55:
    print("ซื้อกาแฟได้ 1 แก้ว")
else:
    print("ดื่มน้ำเปล่าไปก่อน")
```

```python
# เครื่องหมายที่ใช้บ่อย: ==  !=  >  <  >=  <=   and  or  not   in   is None
weather = "rain"
temperature = 31

if weather == "rain" or temperature > 35:
    print("อยู่บ้านดีกว่า")
if weather in ["rain", "storm"]:
    print("พกร่ม")
if note is None:
    print("ไม่มีโน้ต")
```

```python
# ค่าว่างถือว่าเป็น "เท็จ":  ""  0  []  {}  None
todo = []
if not todo:
    print("วันนี้ว่าง!")
```

## 1.8 Loop: ทำซ้ำ

- `for` : ทำกับทุกชิ้นในรายการ
- `while` : ทำซ้ำไปเรื่อย ๆ จนเงื่อนไขเป็นเท็จ หรือเจอ `break`

```python
for item in shopping_list:
    print("ซื้อ", item)
```

```python
for i, item in enumerate(shopping_list, start=1):     # ได้ลำดับมาด้วย
    print(i, item)

for day in range(1, 4):                               # range(1, 4) = 1, 2, 3
    print("วันที่", day)
```

```python
# รวมราคาออเดอร์ด้วย loop
order_total = 0
for item in order["items"]:
    order_total += item["qty"] * item["price"]       # += คือ บวกเพิ่มเข้าไป
print("ออเดอร์นี้", order_total, "บาท")
```

```python
# while: เก็บเงินวันละ 50 จนกว่าจะพอซื้อหูฟัง 400 บาท
savings = 0
days = 0
while savings < 400:
    savings += 50
    days += 1
print(f"ใช้เวลา {days} วัน มีเงิน {savings} บาท")
```

```python
# while True + break: ทำไปเรื่อย ๆ แล้วออกเมื่อถึงหน้าสุดท้าย (เดี๋ยวใช้ตอนดึงข้อมูลจากเว็บทีละหน้า)
page = 1
last_page = 3
while True:
    print("อ่านหน้า", page)
    if page >= last_page:
        break
    page += 1
```

## 1.9 สร้าง list ใหม่แบบสั้น (list comprehension)

รูปแบบ `[ค่าใหม่ for x in รายการ if เงื่อนไข]`

```python
prices = [45, 12, 38, 55]
prices_with_vat = [p * 1.07 for p in prices]
cheap = [p for p in prices if p < 40]
print(prices_with_vat)
print(cheap)

photos = ["cat.jpg", "note.txt", "sea.jpg"]
print([f for f in photos if f.endswith(".jpg")])
```
