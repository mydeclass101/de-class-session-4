## ข้อ 9 · อ่าน decorator

**อย่าเพิ่งรัน** cell ตรวจคำตอบ อ่านโค้ดใน cell นั้นแล้วเดาว่า `printed` จะมีอะไรบ้างตามลำดับ
ใส่คำตอบใน `my_guess` เป็น list ของข้อความ แล้วค่อยรันตรวจ

```python
my_guess = [...]
```

```python
# ตรวจคำตอบ (รันหลังจากเดาแล้ว)
printed = []


def gift_wrap(func):
    def wrapper(*args, **kwargs):
        printed.append("open box")
        result = func(*args, **kwargs)
        printed.append("close box")
        return f"🎁 {result}"
    return wrapper


@gift_wrap
def bake(cake):
    printed.append(f"bake {cake}")
    return cake


printed.append(bake("brownie"))
print(printed)
assert my_guess == printed, f"ที่เดาไว้: {my_guess}"
print("✅ ข้อ 9 ผ่าน")
```
